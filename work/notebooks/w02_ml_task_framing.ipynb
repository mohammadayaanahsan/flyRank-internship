## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

## 1. My lane as an ML task (type)

This is a **binary classification / scoring** problem. Each content
page gets scored with a probability of needing review e.g. "likely
declining"), and those scores are then converted into a ranked list.
So the underlying ML task is classification, and the output is used
as a ranking (scoring) tool for a human reviewer.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

## 2. Target or proxy

The target is a **proxy label**: `is_declining_label = (trend_direction == "down")`.

This label comes from a rule applied to the current data window
(current trend direction), not from a genuinely future outcome. It
is useful as a starting point because it is observable and easy to
compute, but it is not a perfect target - a stronger version would
define decline using a future window (e.g. features from the prior
90 days predicting a drop over the next 30 days).

## 3. Success metric

*One metric you can defend. What number means 'good'?*

## 3. Success metric

The metric is **Precision@50** (and Precision@20). A content review
team can only act on a limited number of pages at a time, so what
matters is not overall accuracy but: of the top 50 pages the model
flags, how many are actually declining? This metric directly
reflects how the output would be used in practice - a ranked review
queue with limited capacity.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

## 5. Why ML beats a fixed rule here

In the starter notebooks, a simple hand-written rule
(stale + visible pages) reached Precision@50 of only 0.240, while a
learned random forest reached about 0.740 on the same task. This
gap shows the real pattern isn't a single clean threshold - it's a
combination of many signals (position, freshness, impressions, word
count, engagement) interacting in ways a fixed if-else rule can't
capture. A model can learn these combinations and their relative
importance directly from data, instead of relying on one person's
guess at the right cutoffs.
