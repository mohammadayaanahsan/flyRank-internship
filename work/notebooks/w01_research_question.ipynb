## 2. The question: decision, action, cost of a wrong call

**Unit of analysis:** one content page (content_id).

**Decision this improves:** which pages the content team should
review first for refresh, out of hundreds of candidates.

**Action someone takes:** a content reviewer opens the top-ranked
pages and decides to update, expand, or leave them as is.

**Cost of a wrong recommendation:** if the model flags a page that
is not actually declining, the reviewer wastes limited review time
on it — and a genuinely declining page with real traffic may get
missed instead. Since review capacity is limited, false positives
directly reduce how many real problems get caught.

**Why data/ML can help:** the starter results already show a
learned model beats a hand-written rule at this ranking task
(Precision@50 well above the baseline), suggesting there are
patterns across many signals (position, freshness, impressions,
word count) that a simple rule can't capture alone.

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

## 4. Careful words: what I can and can't claim


What this work CAN say: which pages *observationally* show patterns
associated with decline (stale + visible, low CTR at a given
position, etc.), and a *directional* ranking of which pages are
worth reviewing first, based on historical patterns.

What this work CANNOT say: it cannot prove that refreshing a page
will cause it to recover — that would require a controlled
experiment, not just observed data. It also cannot claim to know
Google's actual ranking algorithm; it only observes outcomes like
position, clicks, and impressions, not the reasons behind them.
Any recommendation is decision-support for a human reviewer, not a
guaranteed fix.
